```python
# ===== CONFIG =====
import os
INPUT_DIR = os.environ.get("ITDA_INPUT_DIR", "./val_images")
OUTPUT_PATH = os.environ.get("ITDA_OUTPUT_PATH", "./submission.csv")
# ==================

CONFIG_PATH = os.environ.get("ITDA_CONFIG_PATH", "configs/baseline_mock.yaml")
RUN_DIR = os.environ.get("ITDA_RUN_DIR", "/tmp/itda_ocr_run")
```

# ITDA OCR Competition — Official Submission Notebook

This notebook is CPU-only and reads evaluator-provided paths. It never downloads weights or calls external APIs at inference time.


```python
import shutil
import sys
from pathlib import Path

ROOT = Path.cwd()
if not (ROOT / "src").exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / "src"))

from ocr_lab.config import load_config
from ocr_lab.pipeline import run_pipeline

config = load_config(ROOT / CONFIG_PATH)
config.setdefault("runtime", {})["device"] = "cpu"

run_dir = Path(RUN_DIR)
run_dir.mkdir(parents=True, exist_ok=True)
metrics = run_pipeline(
    config=config,
    input_dir=INPUT_DIR,
    output_dir=run_dir,
)

output_path = Path(OUTPUT_PATH)
output_path.parent.mkdir(parents=True, exist_ok=True)
shutil.copyfile(run_dir / "predictions.csv", output_path)
print(f"Processed {metrics['image_count']} images; wrote {output_path}")
```
